#### Q.Find orange cap winner of every season


```python

deliveries= pd.read_csv('../Datasets/deliveries.csv')
matches= pd.read_csv('../Datasets/matches.csv')

merged_dataset=deliveries.merge(matches,left_on='match_id',right_on='id')
a=merged_dataset.groupby(['season','batsman'])['batsman_runs'].sum()
a=a.reset_index() # converting back to data frames
a=a.sort_values('batsman_runs',ascending=False)
a.drop_duplicates(subset=['season']).sort_values('season')
```

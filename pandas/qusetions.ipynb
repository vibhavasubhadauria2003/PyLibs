#### Q.Find most districtive batsman in history of IPL
#####              Best StikeRate sr=(number_of_runs/number_of_balls )*100
#####             Batsman must have played 200 balls in 16-20 overs

```python
mask=data['over']>=16
data_death_overs=data[mask]


bats=data_death_overs.groupby('batsman')
batsmansBalls=bats['batsman_runs'].count().sort_values(ascending=False)

mask2=batsmansBalls>=200
batsmansBalls=batsmansBalls[mask2]


batsmanRuns=bats['batsman_runs'].sum().sort_values(ascending=False)
sr=(batsmanRuns/batsmansBalls)*100

sr.sort_values(ascending=False).head(10)
```

#### Same question using isin

```python
mask=data['over']>=16
data_death_overs=data[mask]
bats=data_death_overs.groupby('batsman')
batsmansBalls=bats['batsman_runs'].count().sort_values(ascending=False)

mask2=batsmansBalls>=200
batsmansBalls=batsmansBalls[mask2]
batsman=batsmansBalls.index.tolist()

mask3=data_death_overs['batsman'].isin(batsman)#compares datasets data with list to filter 
final=data_death_overs[mask3].groupby('batsman')
final_runs=final['batsman_runs'].sum().sort_values(ascending=False)
final_balls=final['batsman_runs'].count().sort_values(ascending=False)

strikeRate=(final_runs/final_balls)*100
strikeRate.sort_values(ascending=False).head(10)
```

#### Q.Find batsman with best average

```python

inings_data=data.drop_duplicates(subset=['match_id','batsman'])
grouped=inings_data.groupby('batsman')
no_inings=grouped['match_id'].count().sort_index(ascending=True)
mask1=no_inings>=15
no_inings=no_inings[mask1]
batsman=no_inings.index.tolist()

mask2=data['batsman'].isin(batsman)
data1=data[mask2]
grouped1=data1.groupby('batsman')
batsman_runs=grouped1['batsman_runs'].sum().sort_index(ascending=True)
batsman_runs
avg=batsman_runs/no_inings
avg.sort_values(ascending=False).head(10)
```

#### Q.Find orange cap winner of every season

```python

deliveries= pd.read_csv('../Datasets/deliveries.csv')
matches= pd.read_csv('../Datasets/matches.csv')

merged_dataset=deliveries.merge(matches,left_on='match_id',right_on='id')
a=merged_dataset.groupby(['season','batsman'])['batsman_runs'].sum()
a=a.reset_index() # converting back to data frames
a=a.sort_values('batsman_runs',ascending=False)
a.drop_duplicates(subset=['season']).sort_values('season')
```

##### pandas.pivot_table() function allows us to create a pivot table to summarize and aggregate data. This function is important when working with large datasets to analyze and transform data efficiently. In this article, we will see some examples to see how it works.

```python
import numpy as np
import pandas as pd
import seaborn as sns

deliveries= pd.read_csv('../Datasets/deliveries.csv')
matches= pd.read_csv('../Datasets/matches.csv')
```

##### Q.Team with numberof 6s in each over

```python
mask=deliveries['batsman_runs']==6
sixes=deliveries[mask]
sixes.head(5)
```

```python
a=sixes.pivot_table(index='over',columns='batting_team',values='batsman_runs',aggfunc='count')
a=a.fillna(0)# Repacing NaN with 0s
a.head(20)
```

```python
sns.heatmap(a)
```

##### Q.Numer of wikets taken in each overs by each team

```python
modified=deliveries.fillna('no_data')
mask=modified['player_dismissed']!='no_data'
modified=modified[mask]
b=modified.pivot_table(index='over',columns='bowling_team',values='player_dismissed',aggfunc='count')
b=b.fillna(0)
b.head(20)
```

```python
sns.heatmap(b)
```
